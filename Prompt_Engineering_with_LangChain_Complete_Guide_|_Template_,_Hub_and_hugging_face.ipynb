`With template` (reusable)

"Translate '{word}' to '{language}'"

```python
from langchain_core.prompts import PromptTemplate
#create a template with placeholder in {curly braces}
template=PromptTemplate.from_template("Translate the word '{word}' to {language}. nothing else")


#Fill in the blanks
prompt =template.format(word="hello",language="Sanskrit")
print("Generated prompt: ",prompt)
```

```python
# Send prompt to the AI model
response = llm.invoke(prompt)

# Print AI response
print("Translation is: ", response.content)
```

```python
#Reuse the same template

for word,lang in [{"thank you","Japanese"},{"water","French"},{"water","sanskrit"},{"water","japanese"},{"water","chinese"}]:
  prompt=template.format(word=word,language=lang)
  response=llm.invoke(prompt)
  print(word,"->",response.content)
```
